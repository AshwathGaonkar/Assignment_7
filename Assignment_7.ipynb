```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample dataset
data = {
    'Student': ['Amit', 'Bhavna', 'Chetan',
                'Diya', 'Esha', 'Farhan'],
    'Maths': [75, 82, 68, 90, 72, 85],
    'Python': [80, 78, 70, 92, 75, 88],
    'Attendance': [85, 90, 78, 95, 80, 92],
    'City': ['Pune', 'Mumbai', 'Pune',
             'Nashik', 'Mumbai', 'Pune']
}

df = pd.DataFrame(data)

# Save dataset as CSV
df.to_csv('students.csv', index=False)

# Load CSV file
df = pd.read_csv('students.csv')

print("Student Dataset:")
print(df)

# 1. Line Plot for Python marks
plt.figure()
plt.plot(df['Student'], df['Python'], marker='o')
plt.title('Python Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. Bar Plot for Maths marks
plt.figure()
plt.bar(df['Student'], df['Maths'])
plt.title('Maths Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Scatter Plot: Maths vs Python
plt.figure()
plt.scatter(df['Maths'], df['Python'])
plt.title('Maths vs Python Marks')
plt.xlabel('Maths Marks')
plt.ylabel('Python Marks')
plt.tight_layout()
plt.show()

# 4. Correlation Matrix and Heatmap
numeric_df = df[['Maths', 'Python', 'Attendance']]
corr = numeric_df.corr()

print("\nCorrelation Matrix:")
print(corr)

plt.figure()
sns.heatmap(corr, annot=True, cmap='coolwarm',
            vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
